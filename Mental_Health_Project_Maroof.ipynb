## Step 1 — Data Extraction & Loading

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Plot style
sns.set_theme(style='darkgrid', palette='husl')
plt.rcParams['figure.dpi'] = 120

print('All libraries imported successfully!')
```

```python
# Load the dataset
df = pd.read_csv('Teen_Mental_Health_Dataset.csv')

print('Dataset loaded successfully!')
print(f'Rows    : {df.shape[0]}')
print(f'Columns : {df.shape[1]}')
print(f'\nColumn Names:')
print(df.columns.tolist())
```

## Step 2 — Data Cleaning

```python
# 2a. check missing values
print('Missing values per column:')
print(df.isnull().sum())
```

```python
# 2b. check Duplicate rows
dupes = df.duplicated().sum()
print(f'Duplicate rows: {dupes}')

if dupes > 0:
    df.drop_duplicates(inplace=True)
    print(f'Duplicates removed! New shape: {df.shape}')
else:
    print('No duplicates found. Dataset is clean!')
```

```python
# 2c. Categorical columns to standardize  (lowercase + strip)
cat_cols = ['gender', 'platform_usage', 'social_interaction_level']
for col in cat_cols:
    df[col] = df[col].str.lower().str.strip()

print('Categorical columns standardized!')

# Unique values check karo
for col in cat_cols:
    print(f'{col}: {df[col].unique().tolist()}')
```

```python
# 2d. New feature: add new age_group column 
def age_group(age):
    if age <= 14:
        return 'Early Teen (13-14)'
    elif age <= 16:
        return 'Mid Teen (15-16)'
    else:
        return 'Late Teen (17-19)'

df['age_group'] = df['age'].apply(age_group)
print('age_group column added!')
print(df['age_group'].value_counts())
```

```python
# 2e. Final cleaned dataset summary
print('--- CLEANED DATASET ---')
print(f'Shape: {df.shape}')
print()
df.describe().round(2)
```

## Step 3 — Data Analysis

```python
# 3a. Depression distribution
dep_counts = df['depression_label'].value_counts()
print('Depression Label Distribution:')
print(f'  Not Depressed (0): {dep_counts[0]} ({dep_counts[0]/len(df)*100:.1f}%)')
print(f'  Depressed     (1): {dep_counts[1]} ({dep_counts[1]/len(df)*100:.1f}%)')
```

```python
# 3b. Social media hours by depression
print('Avg Daily Social Media Hours by Depression:')
print(df.groupby('depression_label')['daily_social_media_hours'].mean().round(2))
```

```python
# 3c. Stress, anxiety, addiction by gender
print('Avg Stress, Anxiety & Addiction by Gender:')
df.groupby('gender')[['stress_level', 'anxiety_level', 'addiction_level']].mean().round(2)
```

```python
# 3d. Platform usage count
print('Platform Usage Count:')
print(df['platform_usage'].value_counts())
```

```python
# 3e. NumPy stats — social media hours
hours = df['daily_social_media_hours'].values
print('NumPy Stats — Daily Social Media Hours:')
print(f'  Mean   : {np.mean(hours):.2f}')
print(f'  Median : {np.median(hours):.2f}')
print(f'  Std Dev: {np.std(hours):.2f}')
print(f'  Min    : {np.min(hours)}')
print(f'  Max    : {np.max(hours)}')
```

```python
# 3f. Correlation with depression label
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()['depression_label'].sort_values(ascending=False)
print('Correlation with Depression Label:')
print(corr.round(3))
```

## Step 4 — Data Visualization

```python
# Chart 1: Depression Distribution (Pie Chart)
fig, ax = plt.subplots(figsize=(6, 5))
labels = ['Not Depressed', 'Depressed']
sizes  = [dep_counts[0], dep_counts[1]]
colors = ['#2ecc71', '#e74c3c']
ax.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
ax.set_title('Depression Label Distribution', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Chart 2: Social Media Hours Distribution (Histogram)
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(df['daily_social_media_hours'], bins=20, kde=True, color='steelblue', ax=ax)
ax.set_title('Daily Social Media Hours Distribution', fontsize=14, fontweight='bold')
ax.set_xlabel('Hours per Day')
ax.set_ylabel('Count')
plt.show()
```

```python
# Chart 3: Social Media Hours vs Depression (Box Plot)
fig, ax = plt.subplots(figsize=(7, 5))
sns.boxplot(data=df, x='depression_label', y='daily_social_media_hours',
            palette=['#2ecc71', '#e74c3c'], ax=ax)
ax.set_title('Social Media Hours vs Depression', fontsize=14, fontweight='bold')
ax.set_xlabel('Depression Label (0=No, 1=Yes)')
ax.set_ylabel('Daily Social Media Hours')
plt.show()
```

```python
# Chart 4: Stress Level by Gender (Violin Plot)
fig, ax = plt.subplots(figsize=(7, 5))
sns.violinplot(data=df, x='gender', y='stress_level', palette='Set2', ax=ax)
ax.set_title('Stress Level Distribution by Gender', fontsize=14, fontweight='bold')
ax.set_xlabel('Gender')
ax.set_ylabel('Stress Level')
plt.show()
```

```python
# Chart 5: Platform Usage Count (Bar Chart)
fig, ax = plt.subplots(figsize=(7, 5))
platform_counts = df['platform_usage'].value_counts()
sns.barplot(x=platform_counts.index, y=platform_counts.values, palette='viridis', ax=ax)
ax.set_title('Platform Usage Distribution', fontsize=14, fontweight='bold')
ax.set_xlabel('Platform')
ax.set_ylabel('Number of Teens')
for i, v in enumerate(platform_counts.values):
    ax.text(i, v + 5, str(v), ha='center', fontweight='bold')
plt.show()
```

```python
# Chart 6: Sleep Hours vs Anxiety Level (Scatter Plot)
fig, ax = plt.subplots(figsize=(8, 5))
colors_map = df['depression_label'].map({0: '#2ecc71', 1: '#e74c3c'})
ax.scatter(df['sleep_hours'], df['anxiety_level'],
           c=colors_map, alpha=0.4, s=30)
ax.set_title('Sleep Hours vs Anxiety Level\n(Green=No Depression, Red=Depressed)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Sleep Hours')
ax.set_ylabel('Anxiety Level')
plt.show()
```

```python
# Chart 7: Correlation Heatmap
fig, ax = plt.subplots(figsize=(10, 7))
corr_matrix = df.select_dtypes(include=np.number).corr()
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, ax=ax, annot_kws={'size': 8})
ax.set_title('Correlation Heatmap', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.show()
```

```python
# Chart 8: Avg Addiction Level by Platform & Gender
fig, ax = plt.subplots(figsize=(8, 5))
grouped = df.groupby(['platform_usage', 'gender'])['addiction_level'].mean().reset_index()
sns.barplot(data=grouped, x='platform_usage', y='addiction_level',
            hue='gender', palette='Set1', ax=ax)
ax.set_title('Avg Addiction Level by Platform & Gender', fontsize=14, fontweight='bold')
ax.set_xlabel('Platform')
ax.set_ylabel('Avg Addiction Level')
ax.legend(title='Gender')
plt.show()
```

## Key Findings Summary

```python
avg_sm_no_dep = df[df['depression_label']==0]['daily_social_media_hours'].mean()
avg_sm_dep    = df[df['depression_label']==1]['daily_social_media_hours'].mean()
avg_sleep     = df['sleep_hours'].mean()
top_platform  = df['platform_usage'].value_counts().idxmax()
high_addiction = df[df['addiction_level'] >= 8].shape[0]

print('============================================')
print('         KEY FINDINGS SUMMARY              ')
print('============================================')
print(f'1. Dataset: {len(df)} teens, {df.shape[1]} features')
print(f'2. Depression rate: {dep_counts[1]/len(df)*100:.1f}% of teens')
print(f'3. Avg social media: {np.mean(hours):.1f} hrs/day')
print(f'4. Depressed teens use SM: {avg_sm_dep:.1f} hrs/day')
print(f'   Non-depressed teens  : {avg_sm_no_dep:.1f} hrs/day')
print(f'5. Avg sleep: {avg_sleep:.1f} hrs')
print(f'6. Most used platform: {top_platform.upper()}')
print(f'7. High addiction (8+): {high_addiction} teens ({high_addiction/len(df)*100:.1f}%)')
print('============================================')
print('       Project Complete! - Maroof Ali      ')
print('============================================')
```
